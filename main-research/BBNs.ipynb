```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
raw_data=pd.read_excel("Clean_excell.xlsx")
raw_data.head(10)
```

```python
precise_data=raw_data.drop(['EWO No','Order Date','Item Name','Delivery Date','Supplier Address','Quality','Cost','Time','Reliability','Delivery','Total'],1)
```

```python
Dic = {}
gc1= pd.DataFrame(data=Dic,index=[0],columns=['Vendor','unc_responsiveness','unc_cooperativeness',
                                             'unc_packaging','unc_concistency','unc_on_time','unc_delivery_perform','unc_fineness',
                                             'unc_length','unc_strength','unc_flexibility','unc_ordr_full','unc_after_sale','unc_total_qty',
                                             'unc_lead','unc_unit_price'])
```

```python
len(precise_data)
```

```python


# Initialize an empty list to store the dictionaries
data_list = []

# Iterate over each row in precise_data
for i in range(len(precise_data)):
    vendor = precise_data.Vendor[i]
    unc_resp = 1 - (precise_data['Responsiveness'][i]) / 5
    unc_Co = 1 - (precise_data['Co_operativeness'][i]) / 5
    unc_pack = 1 - (precise_data['Packaging'][i]) / 5
    unc_conc = 1 - (precise_data['Consistency'][i]) / 5
    unc_ont = 1 - precise_data['On time Delivery'][i]
    unc_dlp = 1 - (precise_data['Delivery Performance'][i]) / 5
    unc_fin = 1 - (precise_data['Fineness'][i]) / 5
    unc_len = 1 - (precise_data['Length(cm)'][i] - 109.8622047) / 244.645795
    unc_stren = 1 - (precise_data['Strength (gm/denier)'][i] - 3.85) / 3.15
    unc_flex = 1 - (5 - precise_data['Flexibility'][i]) / 5
    unc_orfl = 1 - (precise_data['Order Fullfillment'][i])
    unc_afss = 1 - (precise_data['After sale servixe'][i]) / 5
    unc_tqik = 1 - (precise_data['Total Qty in (Kg)'][i]) / 1000
    unc_lt = (precise_data['Lead Time'][i]) / 25
    unc_up = (precise_data['Unit price'][i]) / 17

    # Create a dictionary for each vendor
    Dic = {
        'Vendor': vendor,
        'unc_responsiveness': unc_resp,
        'unc_cooperativeness': unc_Co,
        'unc_packaging': unc_pack,
        'unc_concistency': unc_conc,
        'unc_on_time': unc_ont,
        'unc_delivery_perform': unc_dlp,
        'unc_fineness': unc_fin,
        'unc_length': unc_len,
        'unc_strength': unc_stren,
        'unc_flexibility': unc_flex,
        'unc_ordr_full': unc_orfl,
        'unc_after_sale': unc_afss,
        'unc_total_qty': unc_tqik,
        'unc_lead': unc_lt,
        'unc_unit_price': unc_up
    }

    # Append the dictionary to the list
    data_list.append(Dic)

# Create a DataFrame from the list of dictionaries
gc1 = pd.DataFrame(data_list)
```

```python
gc1.to_excel("uncertainty.xlsx")
```

```python
df=pd.ExcelFile("uncertainty2.xlsx")
df1 = pd.read_excel(df, 'Main_Factor')
df2 = pd.read_excel(df, 'Operational_risk_probability')
df3= pd.read_excel(df, 'Network_risk_probability')
df4=pd.read_excel(df, 'External_risk_probability')
```

```python
Dic = {}
gc1= pd.DataFrame(data=Dic,index=[0],columns=['Vendor','Operational_risk_probability'])
gc2= pd.DataFrame(data=Dic,index=[0],columns=['Vendor','Network_risk_probability'])
gc3= pd.DataFrame(data=Dic,index=[0],columns=['Vendor','External_risk_probability'])
gc=pd.DataFrame(data=Dic,index=[0],columns=['Vendor','Operational_risk_probability','Network_risk_probability','External_risk_probability','Total_risk_probability'])
```

```python
for j in range(len(df1)):
    vendor=df1.Vendor[j]
    orp=0
    for i in range(len(df2)):
        if df2.Fineness[i]==1:
            p_a= df1['unc_fineness'][j]
        else:
            p_a=1-df1['unc_fineness'][j]
        if df2.Length[i]==1:
            p_b=df1['unc_length'][j]
        else:
            p_b=1-df1["unc_length"][j]
        if df2.Strength[i]==1:
            p_c=df1['unc_strength'][j]
        else:
            p_c=1-df1['unc_strength'][j]
        if df2['Unit Price'][i]==1:
            p_d=df1['unc_unit_price'][j]
        else:
            p_d=1-df1['unc_unit_price'][j]
        if df2['Total Quantity'][i]==1:
            p_e=df1['unc_total_qty'][j]
        else:
            p_e=1-df1['unc_total_qty'][j]
        if df2['Packaging'][i]==1:
            p_f=df1['unc_packaging'][j]
        else:
            p_f=1-df1['unc_packaging'][j]
        orp=orp+df2['Operational risk'][i]*p_a*p_b*p_c*p_d*p_e*p_f
    Dic={'Vendor':vendor,'Operational_risk_probability':orp}
    gc1 = gc1.append(Dic, ignore_index=True)
print(gc1)
```

```python
gc1.to_excel("Operational_risk.xlsx")
```

```python
gc2.to_excel("Network_risk.xlsx")
```
